````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
# 1
cities_clean = [city.strip().title() for city in cities]

print(cities_clean)

# 2 
amounts_int = []
for amount in amounts:
    amounts_int.append(int(amount))

print(amounts_int)

# 3 
total_orders = len(amounts_int)
total_sum = sum(amounts_int)
average = round(total_sum / total_orders)

print("Количество заказов:", total_orders)
print("Общая сумма:", total_sum)
print("Средний чек:", average)

# 4 
unpaid = [amount for amount, status in zip(amounts_int, paid) if not status]

print("Неоплаченных заказов:", len(unpaid))
print("Сумма неоплаченных:", sum(unpaid))

# 5
print(f"{'Город':<15} | {'Сумма':<10} | {'Статус'}")
print("-" * 40)

for city, amount, status in zip(cities_clean, amounts_int, paid):
    if status:
        result = "Оплачено"
    else:
        result = "Не оплачено"

    print(f"{city:<15} | {amount:<10} | {result}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
# 6
almaty_sum = 0
astana_sum = 0

for city, amount, status in zip(cities_clean, amounts_int, paid):
    if status == True:
        if city == "Almaty":
            almaty_sum += amount
        elif city == "Astana":
            astana_sum += amount

print("Выручка Almaty:", almaty_sum)
print("Выручка Astana:", astana_sum) 

# 7 
max_amount = 0
max_index = -1
max_city = ""

for i, (city, amount, status) in enumerate(zip(cities_clean, amounts_int, paid)):
    if status and amount > max_amount:
        max_amount = amount
        max_index = i
        max_city = city

print("Индекс:", max_index)
print("Город:", max_city)
print("Сумма:", max_amount)

# 8 
weekend_count = 0

for day in days:
    if day == "сб" or day == "вс":
        weekend_count += 1

percent = weekend_count / len(days) * 100

print("Заказов в выходные:", weekend_count)
print("Доля:", round(percent, 2), "%")

# 9.1

money = 0
months = 0

while money < 400000:
    money += 15000
    months += 1

print("Месяцев:", months)
print("Накоплено:", money)

# 9.2 

money = 0
months = 0
save = 15000

while money < 400000:
    money += 15000
    months += 1
    save += 1000 

print("Месяцев:", months)
print("Накоплено:", money)
````
